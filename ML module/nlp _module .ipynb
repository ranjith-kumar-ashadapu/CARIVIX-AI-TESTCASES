```python
#stemming and lemmatization
import nltk
import time
from nltk.corpus import movie_reviews
from nltk.stem import PorterStemmer, WordNetLemmatizer
from nltk.tokenize import word_tokenize
```

```python
nltk.download('movie_reviews')
nltk.download('wordnet')
```

```python
#intializing stemming and lemmatization
stemmer = PorterStemmer()
lemmatizer = WordNetLemmatizer()

def apply_stemming(tokens):
  return [stemmer.stem(token) for token in tokens]

def apply_lemmatization(tokens):
  return [lemmatizer.lemmatize(token) for token in tokens]
```

```python
'''
using the inbuilt nltk movie reviews dataset
'''
sample_ids = movie_reviews.fileids()[:3]
texts = [" ".join(movie_reviews.words(fileid)) for fileid in sample_ids]
results = []
```

```python
#comparing the alg performance
for text in texts:
  tokens = word_tokenize(text)

  #measure stemming time
  start_time = time.time()
  stems = apply_stemming(tokens)
  stem_time = time.time() - start_time

  #measure lemmatization time
  start_time = time.time()
  lemmas = apply_lemmatization(tokens)
  lemma_time = time.time() - start_time

  results.append({
      "Original": text[:200],
      "Tokens": tokens[:20],
      "Stems": stems[:20],
      "Lemmas": lemmas[:20],
      "Stem_Time": stem_time,
      "Lemma_Time": lemma_time
  })
```

```python
#results for documenting
for r in results:
  print("\nPreprocessing Report")
  print("Original Text (sample):", r["Original"])
  print("Tokens (first 20):", r["Tokens"])
  print("Stemming Result:", r["Stems"], f"(Time: {r['Stem_Time']:.6f}s)")
  print("Lemmatization Result:", r["Lemmas"], f"(Time: {r['Lemma_Time']:.6f}s)")
```

```python
#Integrating preprocessing pipeline
import re
import string
from nltk.corpus import stopwords
from nltk.stem import WordNetLemmatizer
from nltk.tokenize import word_tokenize
```

```python
stop_words = set(stopwords.words("english"))
lemmatizer = WordNetLemmatizer()
```

```python
def preprocessing_pipeline(text):
  #lowercase
  text = text.lower()
  #remove number
  text = re.sub(r'\d+', '', text)
  #remove punctuation
  text = text.translate(str.maketrans('', '', string.punctuation))
  #remove extra white spaces
  text = " ".join(text.split())
  #tokenization
  tokens = word_tokenize(text)
  #remove stopwords
  tokens = [word for word in tokens if word not in stop_words]
  #lemmatization
  tokens = [lemmatizer.lemmatize(word) for word in tokens]

  return tokens
```

```python
#testing pipeline
sample = """
Carivix AI is developing intelligent voice assitants for healthcare, finance and customer support. This module processes speech, understands the human language, and generates the response in real-time.
"""

processed = preprocessing_pipeline(sample)

print(f"Processed Tokens:\n {processed}")
```

```python
#testing with sample datasets
#dataset 1 - gutenberg
from nltk.corpus import gutenberg

sample_text = gutenberg.raw('austen-emma.txt')[:500]

processed = preprocessing_pipeline(sample_text)

print(f"Original text: {len(word_tokenize(sample_text))}")
print(f"Processed Text: {processed}")
```

```python
#dataset 2 - movie_reviews
from nltk.corpus import movie_reviews

fileid = movie_reviews.fileids()[0]

review = " ".join(movie_reviews.words(fileid))

processed_review = preprocessing_pipeline(review)

print(f"Original text: {len(word_tokenize(review))}")
print(f"Processed Text: {processed_review}")
```

```python
#measuring preprocessing accuracy
import time

def evaluate_preprocessing(text):
  original_tokens = word_tokenize(text)
  start_time = time.time()
  processed_tokens = preprocessing_pipeline(text)
  end_time = time.time()
  reduction = len(original_tokens) - len(processed_tokens)
  reduction_percentage = (reduction / len(original_tokens)) * 100

  return {
      "Original Tokens": len(original_tokens),
      "Processed Tokens": len(processed_tokens),
      "Removed Tokens": reduction,
      "Reduction %": round(reduction_percentage,2),
      "Execution Time (seconds)": round(end_time -start_time,5)
    }
```

```python
#evaluation on gutenberg dataset
gutenberg_result = evaluate_preprocessing(sample_text)
print(gutenberg_result)
```

```python
#evaluation of movie_reviews dataset
movie_reviews_result = evaluate_preprocessing(review)
print(movie_reviews_result)
```

```python
import numpy as np
import pandas as pd
from nltk.corpus import movie_reviews
from sklearn.feature_extraction.text import CountVectorizer
```

```python
#loading the dataset
documents = []
for fileid in movie_reviews.fileids():
  documents.append(movie_reviews.raw(fileid))

print(f"Total documents: {len(documents)}")
```

```python
processed_documents = []

for doc in documents:
  tokens = preprocessing_pipeline(doc)
  processed_documents.append(" ".join(tokens))
```

```python
print(f"Original document:\n {documents[0][:500]}")
print(f"\nProcessed document:\n {processed_documents[0][:500]}")
```

```python
vectorizer = CountVectorizer()
bow_matrix = vectorizer.fit_transform(processed_documents)
print(f"Document matrix shpae:\n {bow_matrix.shape}")
print(f"Vocabulary size:\n {len(vectorizer.vocabulary_)}")
vocabulary = vectorizer.get_feature_names_out()
print(vocabulary[:50])
```

```python
bow_df = pd.DataFrame(bow_matrix.toarray(), columns=vocabulary)
bow_df.head()
```

```python
word_frequency = np.asarray(bow_matrix.sum(axis=0)).flatten()
frequency_df = pd.DataFrame({"Word": vectorizer.get_feature_names_out(), "Frequency": word_frequency})
frequency_df = frequency_df.sort_values(by="Frequency",ascending=False)
frequency_df.head(20)
```
